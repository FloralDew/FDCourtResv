## 2. 浏览器相关操作
### 2.1 创建、设置、打开

```python
from selenium import webdriver # 用于操作浏览器
from selenium.webdriver.edge.options import Options # 用于设置Edge浏览器
from selenium.webdriver.edge.service import Service

browser_option = Options() # 创建设置浏览器对象
browser_option.add_argument('--no-sandbox') # 禁用沙盒模式. 有的电脑不写这一句浏览器会闪退. 这句代码可以提高兼容性
browser_option.add_experimental_option('detach', True) # 语句执行完毕后, 保持浏览器开启. 默认关闭

# 创建并启动浏览器
driver = webdriver.Edge(service=Service(r"./MsEdgeDriver/msedgedriver.exe"), options=browser_option) # 第一个参数为驱动位置
```

### 3.2 隐性等待
打开网站后有时不能立刻找到元素, 需要给网站加载时间. 这种现象在小网站中很普遍.

```python
driver.implicitly_wait(10) # 设置查找元素最多等待10s. 一旦找到了元素, 退出等待; 10s内未找到, 报错.
search_botton = driver.find_element(By.CSS_SELECTOR, '#search_icon > svg')

# 注意: 之后再次查找不需要加browser.implicitly_wait(10), 这是一个设置语句, 只用写一遍
```

### 3.3 八种定位方式
#### 3.3.1 ID(之前演示过)
- 好处: 一般比较准确, 重复的比较少.
- 坏处: 但是有的元素没有ID
#### 3.3.2 NAME
- 优势劣势与ID定位相同.
- NAME出现的频率比ID低
#### 3.3.3 CLASS_NAME
- 定位时class值**不能有空格**(Compound class is not allowed), 否则报错. class值的空格表示此元素属于多个class, 挑选其中一个便可以用来定位
- 大部分元素都有class值, 这使得这种查找方式很泛用; 也使得重复的有很多, 因此要使用列表索引选择指定元素. (如何确定索引? 在控制台中输入`document.getElementByClassName("xxx")`后, 鼠标悬浮找元素)
- 部分网站ClassName随机, 难以定位
#### 3.3.4 TAG_NAME
- 指标签的名字. 即&lt;input ...>&lt;span ...>里的"input", "span"
- 重复特别多
#### 3.3.5 LINK_TEXT
- 只针对TAG_NAME为"a"的标签(超链接), 通过其文本内容精确查找(通常是显示在网页上的文本内容)
- 有时&lt;a>没有文本, 而是一个图片, 此时这种查找方法失效
#### 3.3.6 PARTIAL_LINK_TEXT
- 与LINK_TEXT相同, 但是模糊查找(只用包含完整文本的一部分)
#### 3.3.7 CSS_SELECTOR
- 样式选择器, 可以使用多种方法

|语法|意义|
|--:|:--|
|#id|通过ID定位|
|.class|通过className定位|
|不加修饰符|通过标签头定位|
|[类型='精确值']|通过任意类型定位|
|[类型*='模糊值']|通过任意类型定位|
|[类型^='开头值']|通过任意类型定位|
|[类型$='结尾值']|通过任意类型定位|
> 模糊值: 只用包含完整值的一部分
- **更简单的方式:** 在Inspect中直接Copy Selector, 粘贴入By.CSS_SELECTOR, "___"中. 并且这种方式一定是唯一的. (缺点: 个别元素的定位值会比较长)

```python
# CSS_SELECTOR
search_box = driver.find_element(By.CSS_SELECTOR, '#sb_form_q') # 通过bing的搜索框的id定位搜索框
print(search_box)
search_box = driver.find_element(By.CSS_SELECTOR, '.sb_form_q') # 通过bing的搜索框的className定位搜索框
print(search_box)
search_box = driver.find_element(By.CSS_SELECTOR, "[maxlength='1000']") # 通过bing的搜索框的maxlength定位搜索框
print(search_box)
```

#### 3.3.8 XPATH (唯一可以处理属性随机的定位方法)
- 写法一: 通过属性+路径定位. 但如果属性随机, 定位不到.
  - 在Inspect中Copy XPATH, 粘贴入By.XPATH, "___"中.
- 写法二: 通过完整路径定位. 几乎可以定位到任何元素. (缺点: 元素的定位值会比较长)
  - 在Inspect中Copy full XPATH, 粘贴入By.XPATH, "___"中. 并且这种方式一定是唯一的.

### 5.2 IFRAME
在网页下嵌套的另一个网页, 需要进入嵌套的网页才能控制其中元素

```python
driver.get("https://sahitest.com/demo/iframesTest.htm")
# 获取这个标签<iframe ...>对象
iframe = driver.find_element(By.XPATH, '/html/body/iframe')
# 进入这个嵌套页面
driver.switch_to.frame(iframe)
# 点击嵌套页面中的元素
driver.find_element(By.XPATH, '/html/body/table/tbody/tr/td[1]/a[1]').click()
# 退出iframe, 回到原网页才能再操作原网页中的元素
driver.switch_to.default_content() # 无论有多少层嵌套, 直接退回到最外层. 如要退回到特定层, 使用swich_to.frame()
```
